## Q6. Median value of horsepower

```python
hp = df['horsepower']
median_before = hp.median()
median_before
```

```python
most_frequent = hp.mode()[0]
most_frequent
```

```python
hp_filled = hp.fillna(most_frequent)
median_after = hp_filled.median()
median_after
```

```python
print(f"Antes: {median_before:.1f}, Despues: {median_after:.1f}")
if median_before == median_after:
    print("No cambio")
elif median_after > median_before:
    print("Yes, it increased")
else:
    print("Yes, it decreased")
```

## Q7. Sum of weights

```python
asia = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
X = asia.to_numpy()
X
```

```python
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
w
```
